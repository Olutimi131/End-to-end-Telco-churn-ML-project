# Telco Customer Churn: Exploratory Data Analysis & Business Insights

## 1. Executive Overview
Customer churn is a major driver of margin erosion in telecommunication businesses. Acquiring a new subscriber is typically 5x to 7x more expensive than retaining an existing one. This exploratory analysis uncovers the core behavioral, contract, and demographic drivers of customer churn using the canonical IBM Telco Customer Churn dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Styling
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["font.size"] = 11
```

## 2. Ingest and Inspect Raw Data

```python
df = pd.read_csv("../data/raw/telco_churn.csv")
print("Dataset Shape:", df.shape)
df.head()
```

```python
# Data cleaning: TotalCharges contains blank spaces for tenure == 0
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"].astype(str).str.strip(), errors="coerce").fillna(0.0)
df.info()
```

## 3. Target Distribution: Churn Rate

```python
churn_counts = df["Churn"].value_counts()
churn_pct = df["Churn"].value_counts(normalize=True) * 100

fig, ax = plt.subplots(figsize=(6, 4))
sns.barplot(x=churn_counts.index, y=churn_counts.values, palette=["#10B981", "#EF4444"], ax=ax)
ax.set_title(f"Target Distribution: Overall Churn Rate = {churn_pct['Yes']:.1f}%")
ax.set_ylabel("Number of Customers")
plt.show()
```

## 4. Key Churn Drivers Analysis

### A. Contract Type & Tenure Impact

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Contract vs Churn
contract_churn = pd.crosstab(df["Contract"], df["Churn"], normalize="index") * 100
contract_churn.plot(kind="bar", stacked=True, color=["#10B981", "#EF4444"], ax=axes[0])
axes[0].set_title("Churn Rate by Contract Type")
axes[0].set_ylabel("Percentage (%)")

# 2. Tenure distribution
sns.kdeplot(data=df, x="tenure", hue="Churn", common_norm=False, fill=True, palette=["#10B981", "#EF4444"], ax=axes[1])
axes[1].set_title("Tenure Density: Churners vs Retained")

plt.tight_layout()
plt.show()
```

**Finding:** Month-to-month contracts have a >42% churn rate compared to <11% for 1-year and <3% for 2-year contracts. High churn occurs predominantly within the first 6-12 months of customer lifecycle.

### B. Billing & Payment Methods

```python
plt.figure(figsize=(10, 4))
pay_churn = pd.crosstab(df["PaymentMethod"], df["Churn"], normalize="index") * 100
pay_churn.plot(kind="barh", stacked=True, color=["#10B981", "#EF4444"])
plt.title("Churn Rate by Payment Method")
plt.xlabel("Percentage (%)")
plt.show()
```

**Finding:** Electronic check users churn at an alarming ~45% rate, whereas Auto-Pay (Bank transfer / Credit card) accounts have <18% churn.

### C. Internet Service & Support Add-ons

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df, x="InternetService", y="MonthlyCharges", hue="Churn", palette=["#10B981", "#EF4444"], ax=axes[0])
axes[0].set_title("Monthly Charges vs Internet Service & Churn")

tech_churn = pd.crosstab(df["TechSupport"], df["Churn"], normalize="index") * 100
tech_churn.plot(kind="bar", stacked=True, color=["#10B981", "#EF4444"], ax=axes[1])
axes[1].set_title("Churn Rate by Tech Support Subscription")
axes[1].set_ylabel("Percentage (%)")

plt.tight_layout()
plt.show()
```

## 5. Strategic Recommendations for Retention

1. **Contract Migration Incentives:** Provide a $10/month credit or free streaming service upgrade to move customers from Month-to-Month to an Annual Contract.
2. **First-Year Onboarding:** Deploy automated health checks and VIP support during the first 90 days.
3. **Auto-Pay Adoption Campaigns:** Offer a one-off $15 bill discount to migrate electronic check customers to automated credit card / bank transfer.
4. **Bundled Tech Support:** Customers with Tech Support churn at half the rate of those without; bundle basic tech support with Fiber Optic packages.
